```python
#9 function (args & **kwargs)

def calculate_total(*prices, tax_rate=0.05, **discounts):
    
    # Add all prices
    total = sum(prices)

    # Apply tax
    total = total + (total * tax_rate)

    # Subtract discounts
    total = total - sum(discounts.values())

    return total


final_price = calculate_total(80, 230, 350, tax_rate=0.05, coupon=10)

print("Final price:", final_price)
```

```python
#10 oop
class Wallet:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        self.history = []

    def add_funds(self, amount):
        self.balance += amount
        self.history.append(f"Added ₹{amount}")

    def pay(self, amount, merchant):
        if amount <= self.balance:
            self.balance -= amount
            self.history.append(f"Paid ₹{amount} to {merchant}")
        else:
            self.history.append(f"Payment failed: ₹{amount} to {merchant}")

    def get_statement(self):
        print("Transaction History:")
        
        for transaction in self.history:
            print(transaction)

        print("Final Balance: ₹", self.balance)


# Create Wallet object
wallet = Wallet("Balu")

# Add funds
wallet.add_funds(1000)

# Make payments
wallet.pay(500, "Amazon")
wallet.pay(300, "Swiggy")

# Print statement
wallet.get_statement()
```

```python
#2 primary demonstrations
#Geometric shape Evaluation

class Shape:
    def calculate_area(self):
        raise NotImplementedError("Subclasses must implement")


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return 3.14159 * (self.radius ** 2)


class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def calculate_area(self):
        return self.w * self.h


# Uniform Processing Routine
shapes = [Circle(5), Rectangle(4, 6), Circle(2)]

for s in shapes:
    print(f"Area: {s.calculate_area():.2f}")
```

```python
#E-commerce payment pipeline

class PaymentProcessor:
    def process(self, amount):
        pass


class CreditCardPayment(PaymentProcessor):
    def process(self, amount):
        print(f"Charged ${amount} via Credit Card.")


class PayPalPayment(PaymentProcessor):
    def process(self, amount):
        print(f"Authorized ${amount} via PayPal.")


def checkout(processor: PaymentProcessor, amount: float):
    processor.process(amount)


# Using different payment methods
checkout(CreditCardPayment(), 100)
checkout(PayPalPayment(), 250)
```

```python
#task2 notifications & alert dispatcher

class NotificationChannel:
    def send_alert(self, message):
        print("Sending notification")


class EmailNotification(NotificationChannel):
    def send_alert(self, message):
        print("Email Alert:", message)


class SMSNotification(NotificationChannel):
    def send_alert(self, message):
        print("SMS Alert:", message)


class SlackNotification(NotificationChannel):
    def send_alert(self, message):
        print("Slack Alert:", message)


class AlertManager:
    def __init__(self, channels):
        self.channels = channels

    def broadcast(self, message):
        for channel in self.channels:
            channel.send_alert(message)


# Create notification channels
email = EmailNotification()
sms = SMSNotification()
slack = SlackNotification()

# Create AlertManager
manager = AlertManager([email, sms, slack])

# Send emergency alert
manager.broadcast("Server is down!")
```

```python
# task3 universal file parser pipeline

from abc import ABC, abstractmethod
import csv
import json
import xml.etree.ElementTree as ET


# Base Abstract Class
class FileParser(ABC):

    @abstractmethod
    def open_file(self):
        pass

    @abstractmethod
    def parse(self):
        pass


# CSV Parser
class CSVParser(FileParser):

    def __init__(self, filename):
        self.filename = filename
        self.file = None

    def open_file(self):
        self.file = open(self.filename, "r")
        print("CSV file opened")

    def parse(self):
        reader = csv.reader(self.file)
        data = list(reader)
        print("CSV data parsed")
        return data


# JSON Parser
class JSONParser(FileParser):

    def __init__(self, filename):
        self.filename = filename
        self.file = None

    def open_file(self):
        self.file = open(self.filename, "r")
        print("JSON file opened")

    def parse(self):
        data = json.load(self.file)
        print("JSON data parsed")
        return data


# XML Parser
class XMLParser(FileParser):

    def __init__(self, filename):
        self.filename = filename
        self.file = None

    def open_file(self):
        self.file = open(self.filename, "r")
        print("XML file opened")

    def parse(self):
        tree = ET.parse(self.file)
        root = tree.getroot()

        data = []
        for item in root:
            data.append(item.text)

        print("XML data parsed")
        return data


# Advanced Batch Processor
def process_data_batch(parsers):

    for parser in parsers:
        parser.open_file()
        data = parser.parse()

        print("Processed Data:", data)
        print("--------------------")


# Create parser objects
parsers = [
    CSVParser("students.csv"),
    JSONParser("students.json"),
    XMLParser("students.xml")
]

```
