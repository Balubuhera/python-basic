```python
# easy
#1

celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = (celsius * 9 / 5) + 32

print("Temperature in Fahrenheit:", fahrenheit)
```

```python
# intermedate
#2
from datetime import date

birth_year = int(input("Enter your birth year: "))

birth_date = date(birth_year, 1, 1)
today = date.today()

years = today.year - birth_date.year
months = today.month - birth_date.month
days = today.day - birth_date.day

if days < 0:
    months -= 1
    days += 30

if months < 0:
    years -= 1
    months += 12

print("Your age is:", years, "years,", months, "months,", days, "days")
```

# for loop and while loop

```python
#1 easy
n = int(input("Enter a number: "))

for i in range(1, 11):
    print(n, "x", i, "=", n * i)
```

```python
#2 intermedate
num = int(input("Enter a positive integer: "))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed number:", reverse)
```

# functions

```python
#1 easy
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


num = int(input("Enter a number: "))

print(is_prime(num))
```

```python
#2 intermedate
def filter_integers(**kwargs):
    result = {}

    for key, value in kwargs.items():
        if isinstance(value, int):
            result[key] = value

    return result


data = filter_integers(name="Balu", age=23, marks=90, city="Chennai")

print(data)
```

# list


```python
#1 easy
numbers = [10, 25, 15, 40, 15]

largest = numbers[0]
smallest = numbers[0]

for n in numbers:
    if n > largest:
        largest = n

    if n < smallest:
        smallest = n

print("Largest:", largest)
print("Smallest:", smallest)
```

```python
#2 intermedate
def move_zeros(numbers):
    result = []

    for num in numbers:
        if num != 0:
            result.append(num)

    for num in numbers:
        if num == 0:
            result.append(num)

    return result


numbers = [0, 1, 0, 3, 12]

print("Original list:", numbers)
print("After moving zeros:", move_zeros(numbers))
```

#  Dictionaries

```python
#1 easy
sentence = input("Enter a sentence: ")

words = sentence.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print("Word frequency:")
print(frequency)
```

```python
#2 intermedate
def merge_dicts(dict1, dict2):
    result = dict1.copy()

    for key, value in dict2.items():
        if key in result:
            result[key] = [result[key], value]
        else:
            result[key] = value

    return result


dict1 = {"a": 10, "b": 20}
dict2 = {"b": 30, "c": 40}

result = merge_dicts(dict1, dict2)

print(result)
```
